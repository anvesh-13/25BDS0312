```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np 
```

```python
df=pd.read_csv("/home/anaconda/Downloads/WineQT.csv")
print(df.head(3))
print(df.tail(3)) 
```

```python
print(df.describe())
pd.isna(df).any() 
```

```python
df['quality'].astype(str).str.isnumeric().value_counts()  
```

```python
target_column="citric acid"
plt.figure(figsize=(10, 6))


sns.histplot(data=df, x=target_column, bins=20, kde=True, color='skyblue', edgecolor='black')


plt.title(f'Histogram Distribution of {target_column}', fontsize=14, fontweight='bold')
plt.xlabel(target_column, fontsize=12)
plt.ylabel('Frequency / Count', fontsize=12)
plt.grid(axis='y', alpha=0.3)


plt.tight_layout()
plt.show()
```

```python
sns.boxplot(data=df, y=target_column, color='lightblue', width=0.4)


plt.title(f'Box Plot of {target_column}', fontsize=14, fontweight='bold')
plt.ylabel(target_column, fontsize=12)
plt.grid(axis='x', alpha=0.3)

plt.tight_layout()
plt.show()
```

```python

numeric_df = df.select_dtypes(include=['number'])
correlation_matrix = numeric_df.corr()
plt.figure(figsize=(10, 8))
sns.heatmap(
    correlation_matrix, 
    annot=True, 
    cmap='coolwarm', 
    fmt=".2f", 
    linewidths=0.5, 
    vmin=-1, 
    vmax=1
)
plt.title('Correlation Heat Map', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))

# 3. Create the bivariate box plot
# x: Categorical grouping variable (e.g., 'Gender', 'Department', 'Year')
# y: Continuous numeric variable (e.g., 'Salary', 'Age', 'Score')
x_column = 'quality'
y_column = 'chlorides'

sns.boxplot(data=df, x=x_column, y=y_column, palette='Set2')

# 4. Add formatting and labels
plt.title(f'Distribution of {y_column} by {x_column}', fontsize=14, fontweight='bold')
plt.xlabel(x_column, fontsize=12)
plt.ylabel(y_column, fontsize=12)
plt.xticks(rotation=45) # Rotates X-axis labels if they overlap
plt.grid(axis='y', alpha=0.3)

# 5. Display the plot
plt.tight_layout()
plt.show()
```

```python
selected_columns = ['density', 'quality', 'fixed acidity']  
df_subset = df[selected_columns]

# 3. Create the pair plot using only the subset
sns.pairplot(data=df_subset, diag_kind='kde', plot_kws={'alpha': 0.6, 'color': 'teal'})

# 4. Display the plot
plt.show()
```

```python
x_axis_column = 'fixed acidity'
y_axis_column = 'pH'

# 3. Create the scatter plot
plt.figure(figsize=(8, 6))
sns.scatterplot(data=df, x=x_axis_column, y=y_axis_column, color='teal', alpha=0.7, s=60)

# 4. Add formatting and labels
plt.title(f'Scatter Plot: {x_axis_column} vs {y_axis_column}', fontsize=14, fontweight='bold')
plt.xlabel(x_axis_column, fontsize=12)
plt.ylabel(y_axis_column, fontsize=12)
plt.grid(True, linestyle='--', alpha=0.5)

# 5. Display the plot
plt.tight_layout()
plt.show()
```
